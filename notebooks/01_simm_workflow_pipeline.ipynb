# Simm Workflow Pipeline

This lightweight notebook uses the importable `rockphys` pipeline. Use `gassmann_demo.ipynb` for the full scientific walkthrough and this notebook when you want a short, repeatable API entry point.

Treat formula changes as scientific decisions. The pipeline preserves the current Simm notebook behavior by default, including facies-based dry-rock conditioning. Change configuration deliberately and pin output changes with tests.

```python
from pathlib import Path
import sys

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from rockphys import SimmWorkflowConfig, run_simm_workflow
```

```python
config = SimmWorkflowConfig(data_dir=ROOT / "data")
result = run_simm_workflow(config)
well = result.well
coeffs = result.coeffs

print(f"rows={len(well)}")
print(f"facies_samples={(well['facies'] != 'Background').sum()}")
print(f"coeffs={coeffs[0]:.3f},{coeffs[1]:.3f},{coeffs[2]:.1f}")
print(f"k_oil={result.oil_bulk_modulus_gpa:.3f} GPa")
```
